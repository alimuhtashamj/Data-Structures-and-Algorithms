```python
# copying an array
num = [1, 2, 3, 4, 5]
new_array = []
for i in num:
    new_array.append(i)
print(new_array)
```

```python
# filtering an array
num = [1, 2, 3, 4, 5]
new_array = []
for i in num:
    if i % 2 == 0:
        new_array.append(i)
print(new_array)
```
