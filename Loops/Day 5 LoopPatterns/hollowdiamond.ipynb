```python
rows = 4
for i in range(1, rows+1):
    for j in range(rows - i):
        print(' ', end = '' )
    print('*', end = '')
    
    if i>1:
        for j in range(2*i-3):
            print(' ', end = '')
        print('*', end = '')
    print()
    
```

```python
rows = 9 
middle = (rows // 2)+1
for i in range(1,middle+1):
    #leading spaces
    for j in range(middle -i):
        print(' ', end = '')
        #after printing leading spaces, 8 is printed afterwards by outerloop.
        #so outer managed the row number and printing of 8.
    print('8', end = '')
    if i>1:
        #printing of inside spaces
        for j in range(2*i-3):
            print(' ', end = '')
        # if conditional prints 8 after the inside spaces by the loop are printed    
        print('8', end = '')
    print()  
 
for i in range(middle-1, 0, -1):
    #bottom half values need to be reversed so the ranges is revered as well
    for j in range(middle-i):
        print(' ', end = '')
    print('8', end = '')
    if i > 1:
        for j in range(2*i-3):
            print(' ', end = '')
        print('8', end = '')
    print()
        
    
        
        
```
