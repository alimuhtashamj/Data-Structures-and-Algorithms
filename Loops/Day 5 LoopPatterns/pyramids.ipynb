```python
#simple pyramid 
rows = 5 
for i in range(1,rows+1):
    for j in range(rows-i):
        print(' ', end = '')
    for k in range(2*i-1):
        print("*" , end = '')
    print()
```

```python
rows = 9
for i in range(1, rows+1):
    for j in range(i-1):
        print(' ', end = '')
    for k in range(2*(rows-i)+1):
        print('8', end = '')
    print()
```
